# SANKEY DIAGRAM FOR PHASE 2.2

```python
import pandas as pd

# Load your exported Sankey-ready CSV
df_sankey = pd.read_csv("sankey_diagram_prep.csv")  # replace with actual file name

# Preview
df_sankey.head()
```

```python
import plotly.graph_objects as go

# Summarize total interventions by gender and year
summary = df_sankey.groupby(["year", "gender"])["n_interventions"].sum().reset_index()

# Build overall totals
total_ints = summary["n_interventions"].sum()
total_f = summary[summary["gender"] == "F"]["n_interventions"].sum()
total_m = summary[summary["gender"] == "M"]["n_interventions"].sum()

# Build year labels with breakdowns
year_labels = []
for year in [2003, 2013, 2023]:
    f_count = summary[(summary["year"] == year) & (summary["gender"] == "F")]["n_interventions"].sum()
    m_count = summary[(summary["year"] == year) & (summary["gender"] == "M")]["n_interventions"].sum()
    label = f"{year} (F: {f_count}, M: {m_count})"
    year_labels.append(label)

# Define node labels
nodes = [
    f"Interventions ({total_ints})",
    f"F ({total_f})",
    f"M ({total_m})",
] + year_labels

# Helper to get index
def idx(label_start):
    return next(i for i, node in enumerate(nodes) if node.startswith(label_start))

# Build links
links = {
    "source": [],
    "target": [],
    "value": [],
    "color": []
}

# Interventions → Gender
for gender, color in [("F", "rgba(161,102,171,0.5)"), ("M", "rgba(70,117,192,0.5)")]:
    val = summary[summary["gender"] == gender]["n_interventions"].sum()
    links["source"].append(idx("Interventions"))
    links["target"].append(idx(gender))
    links["value"].append(val)
    links["color"].append(color)

# Gender → Year
for _, row in summary.iterrows():
    gender = row["gender"]
    year = str(row["year"])
    val = row["n_interventions"]
    color = "rgba(161,102,171,0.5)" if gender == "F" else "rgba(70,117,192,0.5)"
    links["source"].append(idx(gender))
    links["target"].append(idx(year))
    links["value"].append(val)
    links["color"].append(color)

# Plot
fig = go.Figure(go.Sankey(
    arrangement="snap",
    node=dict(
        pad=15,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=nodes,
        color=["#999999", "#A166AB", "#4675C0", "#CCCCCC", "#CCCCCC", "#CCCCCC"]
    ),
    link=dict(
        source=links["source"],
        target=links["target"],
        value=links["value"],
        color=links["color"]
    )
))


fig.update_layout(
    title_text="Interventions → Gender → Year (with totals)", 
    font_size=16, 
    height=700,

)
fig.show()
```

```python
import pandas as pd
import plotly.graph_objects as go

# Load the data
df_freq = pd.read_csv("frequent_speakers.csv")

# Classify as active/inactive (more than 1 speech = active)
df_freq["freq_status"] = df_freq["n_interventions"].apply(lambda x: "Active" if x > 1 else "Inactive")

# Count MPs per group
summary = df_freq.groupby(["freq_status", "gender"])["speaker_id"].nunique().reset_index(name="n_mps")

# Totals
total_mps = df_freq["speaker_id"].nunique()
total_f = df_freq[df_freq["gender"] == "F"]["speaker_id"].nunique()
total_m = df_freq[df_freq["gender"] == "M"]["speaker_id"].nunique()

active_count = df_freq[df_freq["freq_status"] == "Active"]["speaker_id"].nunique()
inactive_count = df_freq[df_freq["freq_status"] == "Inactive"]["speaker_id"].nunique()

# Update node labels
node_labels = [
    f"MPs ({total_mps})",        # 0
    f"Active ({active_count})",  # 1
    f"Inactive ({inactive_count})",  # 2
    f"F ({total_f})",            # 3
    f"M ({total_m})"             # 4
]

# Internal reference map
label_keys = ["MPs", "Active", "Inactive", "F", "M"]
label_to_idx = {k: i for i, k in enumerate(label_keys)}

# Build links
links = {
    "source": [],
    "target": [],
    "value": [],
    "color": []
}

# MPs → Frequency layer
for status in ["Active", "Inactive"]:
    count = df_freq[df_freq["freq_status"] == status]["speaker_id"].nunique()
    links["source"].append(label_to_idx["MPs"])
    links["target"].append(label_to_idx[status])
    links["value"].append(count)
    links["color"].append("rgba(160,160,160,0.4)")

# Frequency → Gender layer
for _, row in summary.iterrows():
    status = "Active" if row["freq_status"] == "Active" else "Inactive"
    gender = row["gender"]
    val = row["n_mps"]
    color = "rgba(161,102,171,0.5)" if gender == "F" else "rgba(70,117,192,0.5)"
    links["source"].append(label_to_idx[status])
    links["target"].append(label_to_idx[gender])
    links["value"].append(val)
    links["color"].append(color)

# Plot
fig = go.Figure(go.Sankey(
    arrangement="snap",
    node=dict(
        pad=15,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=node_labels,
        color=["#999999", "#CCCCCC", "#CCCCCC", "#A166AB", "#4675C0"]
    ),
    link=dict(
        source=links["source"],
        target=links["target"],
        value=links["value"],
        color=links["color"]
    )
))

fig.update_layout(
    title_text="MPs → Active/Inactive Speakers → Gender<br><span style='font-size:11px'><i>Note: Active = spoke more than once (excludes oath)</i></span>",
    font_size=16,
    height=650
)
fig.show()
```

```python

# Extract term from speaker_id (after the dot)
df_freq["term"] = df_freq["speaker_id"].astype(str).str.split(".").str[-1]

# Map term to year
term_to_year = {
    "VII": 2003,
    "X": 2013,
    "XIV": 2023,
    "XV": 2023
}
df_freq["year"] = df_freq["term"].map(term_to_year)

# Classify frequency
df_freq["freq_status"] = df_freq["n_interventions"].apply(lambda x: "Active" if x > 1 else "Inactive")
```

```python
def sankey_by_year(year):
    df_year = df_freq[df_freq["year"] == year].copy()

    # Build summaries
    total_mps = df_year["speaker_id"].nunique()
    total_f = df_year[df_year["gender"] == "F"]["speaker_id"].nunique()
    total_m = df_year[df_year["gender"] == "M"]["speaker_id"].nunique()
    active_df = df_year[df_year["freq_status"] == "Active"]
    inactive_df = df_year[df_year["freq_status"] == "Inactive"]
    active_count = active_df["speaker_id"].nunique()
    inactive_count = inactive_df["speaker_id"].nunique()

    # % helpers
    p = lambda x: f"{round((x / total_mps) * 100)}%"
    pf = lambda df: round((df[df["gender"] == "F"]["speaker_id"].nunique() / df["speaker_id"].nunique()) * 100) if df["speaker_id"].nunique() > 0 else 0

    # % of women in each group
    active_f_pct = pf(active_df)
    inactive_f_pct = pf(inactive_df)

    # Node labels with % + female share
    node_labels = [
        f"MPs (100%)",                            # 0
        f"Active ({p(active_count)}, {active_f_pct}% women)",     # 1
        f"Inactive ({p(inactive_count)}, {inactive_f_pct}% women)",  # 2
        f"F ({p(total_f)})",                      # 3
        f"M ({p(total_m)})"                       # 4
    ]
    label_keys = ["MPs", "Active", "Inactive", "F", "M"]
    label_to_idx = {k: i for i, k in enumerate(label_keys)}

    # Gender breakdown within status
    summary = df_year.groupby(["freq_status", "gender"])["speaker_id"].nunique().reset_index(name="n_mps")

    # Build links
    links = {"source": [], "target": [], "value": [], "color": []}

    for status in ["Active", "Inactive"]:
        count = df_year[df_year["freq_status"] == status]["speaker_id"].nunique()
        links["source"].append(label_to_idx["MPs"])
        links["target"].append(label_to_idx[status])
        links["value"].append(count)
        links["color"].append("rgba(160,160,160,0.4)")

    for _, row in summary.iterrows():
        status = row["freq_status"]
        gender = row["gender"]
        val = row["n_mps"]
        color = "rgba(161,102,171,0.5)" if gender == "F" else "rgba(70,117,192,0.5)"
        links["source"].append(label_to_idx[status])
        links["target"].append(label_to_idx[gender])
        links["value"].append(val)
        links["color"].append(color)

    # Plot
    fig = go.Figure(go.Sankey(
        arrangement="snap",
        node=dict(
            pad=15,
            thickness=20,
            line=dict(color="black", width=0.5),
            label=node_labels,
            color=["#999999", "#CCCCCC", "#CCCCCC", "#A166AB", "#4675C0"]
        ),
        link=dict(
            source=links["source"],
            target=links["target"],
            value=links["value"],
            color=links["color"]
        )
    ))

    fig.update_layout(
        title_text=f"MPs → Active/Inactive Speakers → Gender ({year})",
        font_size=13,
        height=650
    )
    fig.show()
```

```python
sankey_by_year(2003)
sankey_by_year(2013)
sankey_by_year(2023)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
# Group + calculate %
grouped = df_freq.groupby(["year", "freq_status", "gender"])["speaker_id"].nunique().reset_index()
totals = df_freq.groupby(["year", "freq_status"])["speaker_id"].nunique().reset_index(name="total")
merged = pd.merge(grouped, totals, on=["year", "freq_status"])
merged["percentage"] = merged["speaker_id"] / merged["total"] * 100

# F - M gap calculation
pivot_gap = merged.pivot(index=["freq_status", "year"], columns="gender", values="percentage").fillna(0)
pivot_gap["gap"] = pivot_gap["F"] - pivot_gap["M"]
pivot_gap = pivot_gap.reset_index()
pivot_gap["label"] = pivot_gap["freq_status"] + " (F% - M%)"

# Pivot for heatmap
gap_matrix = pivot_gap.pivot(index="label", columns="year", values="gap")
gap_matrix = gap_matrix.reindex(["Active (F% - M%)", "Inactive (F% - M%)"])

# Plot
plt.figure(figsize=(8, 3))
sns.heatmap(gap_matrix, annot=True, fmt=".1f", cmap="RdBu_r", center=0, cbar_kws={"label": "Gender Gap (F% - M%)"})
plt.title("Gender Gap in Symbolic Participation (F% - M%)")
plt.ylabel("Speaker Frequency")
plt.xlabel("Year")
plt.tight_layout()
plt.show()
```

```python
# Group counts
grouped = df_freq.groupby(["year", "freq_status", "gender"])["speaker_id"].nunique().reset_index()
totals = df_freq.groupby(["year", "freq_status"])["speaker_id"].nunique().reset_index(name="total")

# Merge and calc %
merged = pd.merge(grouped, totals, on=["year", "freq_status"])
merged["percentage"] = merged["speaker_id"] / merged["total"] * 100
merged["label"] = merged["freq_status"] + " – " + merged["gender"]

# Pivot for heatmap
heatmap_pct = merged.pivot(index="label", columns="year", values="percentage")
row_order = ["Active – F", "Active – M", "Inactive – F", "Inactive – M"]
heatmap_pct = heatmap_pct.reindex(row_order)

# Plot % heatmap
plt.figure(figsize=(8, 5))
sns.heatmap(heatmap_pct, annot=True, fmt=".1f", cmap="Purples", cbar_kws={"label": "% of MPs"})
plt.title("MP Gender Share by Year and Frequency")
plt.ylabel("Speaker Category")
plt.xlabel("Year")
plt.tight_layout()
plt.show()

```
